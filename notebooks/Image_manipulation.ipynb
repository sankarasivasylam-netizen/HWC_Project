# Data Augemntation - Image Manipulation

2. Non- Geometric Transformations (Cropping, Noise injection, Color Space, Jitter, Kernel)

### Create augmentation pipeline

```python
from torchvision import transforms
```

```python
augmentation_pipeline = transforms.Compose([
    # Geometric Transformations
    transforms.Rotate(limit=25, p=0.5),  # rotation +-25 with prob 0.5
    transforms.ShiftScaleRotate(shift_limit=0.3, scale_limit=0.2, rotate_limit=0, p=0.3),  # translation (shift and scale but do not rotate more)
    transforms.Affine(shear=5, p=0.5),  # shearing up to 5deg - does not seem that useful for us
    transforms.HorizontalFlip(p=0.5),  # flipping h

    # Non-Geometric Transformations
    transforms.RandomResizedCrop(size=[524,524], scale=(0.7, 1.0), p=0.5),  # gets randomly 70-100% of image and resized it back to 256x256
    transforms.GaussNoise(var_limit=(2, 10), p=0.2),  # less variance
    transforms.CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=0.2), # contrast for color (insetad of b&w)
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05, p=0.3),  # Color Adjustments different lightning conditions
    transforms.GaussianBlur(blur_limit=(3, 7), p=0.2),  # Kernel Blurring

    transforms.Resize(524, 524),

    ToTensorV2()
])
```
