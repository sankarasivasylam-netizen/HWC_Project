```python
directory_name_1 = "Organic"
directory_name_2 = "Non-Organics_Recyclables"

# Create the directory
try:
    os.mkdir(directory_name_1)
    print(f"Directory '{directory_name_1}' created successfully.")
except FileExistsError:
    print(f"Directory '{directory_name_1}' already exists.")
except PermissionError:
    print(f"Permission denied: Unable to create '{directory_name_1}'.")
except Exception as e:
    print(f"An error occurred: {e}")

try:
    os.mkdir(directory_name_2)
    print(f"Directory '{directory_name_2}' created successfully.")
except FileExistsError:
    print(f"Directory '{directory_name_2}' already exists.")
except PermissionError:
    print(f"Permission denied: Unable to create '{directory_name_2}'.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import shutil
from pathlib import Path

target_source = Path(".")
target_parent = Path(".")

organic_dir = target_parent / "Organic"
norganic_dir = target_parent / "Non-Organic_Recylables"

organic_dir.mkdir(parents=True, exist_ok=True)
norganic_dir.mkdir(parents=True, exist_ok=True)

organic_set = {"biological"}

def is_organic(folder_name: str) -> bool:
    return folder_name.lower() in organic_set

for waste_folder in target_source.iterdir():
    print(waste_folder)
    if not waste_folder.is_dir() or waste_folder.name == 'Organic' or\
           waste_folder.name == 'Non-Organic_Recylables' : 
        continue

    target_dir = organic_dir if is_organic(waste_folder.name) else norganic_dir
    
    for file in waste_folder.iterdir():
        print(file)
        print(file.name)
        if file.is_file():
            shutil.copy2(file, target_dir / file.name)
```

```python
# Current directory
dataset_path = "."

# List folders (each folder = class)
classes = [d for d in os.listdir(dataset_path) if os.path.isdir(os.path.join(dataset_path, d))]
print("Classes found:", classes)
print("Number of classes:", len(classes))

for cls in classes:
    if cls != 'Organic' and cls != 'Non-Organic_Recylables':
        cls_path = os.path.join(dataset_path, cls)
        num_images = len([f for f in os.listdir(cls_path) if os.path.isfile(os.path.join(cls_path, f))])
        print(f"{cls},{num_images}")
```
