```python
import os
import imagehash
from PIL import Image
from tqdm import tqdm
from pathlib import Path
import pandas as pd
```

```python
def find_duplicates(image_dir):
    
    hashes = {}
    duplicates = []

    for root, dirs, files in os.walk(image_dir):
        for file in files:
            
            if file.lower().endswith(('.png','.jpg','.jpeg','.bmp','.tiff')):
                
                path = os.path.join(root, file)

                try:
                    img = Image.open(path)
                    img_hash = imagehash.phash(img)

                    if img_hash in hashes:
                        duplicates.append((path, hashes[img_hash]))
                    else:
                        hashes[img_hash] = path

                except Exception as e:
                    print("Error reading:", path)

    return duplicates
```

```python
stage2_root = BASE_DIR / "data" / "consolidated" / "Stage-2"

all_duplicates = []
summary = []

for class_folder in os.listdir(stage2_root):

    class_path = os.path.join(stage2_root, class_folder)

    if os.path.isdir(class_path):

        print(f"Checking {class_folder}...")

        duplicates = find_duplicates(class_path)

        summary.append({
            "class": class_folder,
            "duplicate_count": len(duplicates)
        })

        for dup in duplicates:
            all_duplicates.append((class_folder, dup[0], dup[1]))
```

```python
duplicates_df = pd.DataFrame(
    all_duplicates,
    columns=["class","duplicate_image","original_image"]
)
```

```python
import os
import shutil
backup_stg2_dup = BASE_DIR / "data" / "consolidated" / "Stage-2-duplicates"

for _, row in duplicates_df.iterrows():

    class_name = row["class"]
    duplicate_path = row["duplicate_image"]

    # Create class backup folder
    class_backup = os.path.join(backup_stg2_dup, class_name)
    os.makedirs(class_backup, exist_ok=True)

    # Destination path
    filename = os.path.basename(duplicate_path)
    dest_path = os.path.join(class_backup, filename)

    try:
        shutil.move(duplicate_path, dest_path)
        print(f"Moved: {duplicate_path} -> {dest_path}")

    except Exception as e:
        print(f"Error moving {duplicate_path}: {e}")
```
